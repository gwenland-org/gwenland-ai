# GwenLand glcuda Wave 125 - N16 fused SwiGLU


```python
import base64, gzip
from pathlib import Path
Path("/kaggle/working/wave125.patch").write_bytes(gzip.decompress(base64.b64decode('')))
import hashlib
import json
import os
from pathlib import Path
import re
import shutil
import statistics
import subprocess
import traceback
import urllib.request
import zipfile

BUILD = "wave125-n16-fused-swiglu-v1"
REPO_URL = "https://github.com/gwenland-org/gwenland-ai.git"
BASE_REV = "5de5be39c0190b9367da18f0f2001e7f40208c92"
PATCH_SHA256 = "8430590cf1497af6b52b42d05b75426067db1c99a9bf3d28dde8b05e4e197ca3"
MODEL_URL = "https://huggingface.co/Qwen/Qwen2.5-0.5B-Instruct-GGUF/resolve/9217f5db79a29953eb74d5343926648285ec7e67/qwen2.5-0.5b-instruct-q8_0.gguf?download=true"
MODEL_BYTES = 675710816
MODEL_SHA256 = "ca59ca7f13d0e15a8cfa77bd17e65d24f6844b554a7b6c12e07a5f89ff76844e"

ROOT = Path("/kaggle/working/wave125")
TREE = ROOT / "repo"
RESULTS = ROOT / "results"
TARGET = ROOT / "target"
MODEL = ROOT / "qwen2.5-0.5b-instruct-q8_0.gguf"
FINAL_ZIP = Path("/kaggle/working/glcuda-t4-wave125-n16-fused-swiglu-results.zip")
RESULTS.mkdir(parents=True, exist_ok=True)


def run(cmd, cwd=None, env=None, timeout=7200, check=True):
    cmd = [str(x) for x in cmd]
    merged = os.environ.copy()
    if env:
        merged.update({str(k): str(v) for k, v in env.items()})
    print("$", " ".join(cmd), flush=True)
    p = subprocess.run(cmd, cwd=cwd, env=merged, text=True, capture_output=True, timeout=timeout)
    print(p.stdout[-20000:], flush=True)
    print(p.stderr[-20000:], flush=True)
    if check and p.returncode:
        raise RuntimeError(f"command failed ({p.returncode}): {cmd}")
    return p


def save(name, process):
    (RESULTS / name).write_text(
        process.stdout + "\n--- STDERR ---\n" + process.stderr,
        encoding="utf-8",
    )


def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for block in iter(lambda: f.read(1024 * 1024), b""):
            h.update(block)
    return h.hexdigest()


def archive():
    with zipfile.ZipFile(FINAL_ZIP, "w", zipfile.ZIP_DEFLATED) as z:
        for path in sorted(RESULTS.rglob("*")):
            if path.is_file():
                z.write(path, path.relative_to(RESULTS))
    print("ARCHIVE", FINAL_ZIP, sha256_file(FINAL_ZIP), flush=True)


def find_patch():
    matches = []
    for root in [Path("/kaggle/input"), Path("/kaggle/working"), Path.cwd()]:
        if root.exists():
            matches.extend(root.rglob("wave125.patch"))
    good = [p for p in matches if p.is_file() and sha256_file(p) == PATCH_SHA256]
    if not good:
        raise RuntimeError(f"uploaded wave125.patch not found; candidates={matches}")
    return good[0]


phase = "bootstrap"
try:
    patch_path = find_patch()
    (RESULTS / "source.json").write_text(
        json.dumps(
            {
                "build": BUILD,
                "base_rev": BASE_REV,
                "patch_sha256": PATCH_SHA256,
                "patch_bytes": patch_path.stat().st_size,
                "uploaded_path": str(patch_path),
            },
            indent=2,
        ),
        encoding="utf-8",
    )
    gpu = run(
        [
            "nvidia-smi",
            "--query-gpu=index,name,compute_cap,memory.total,driver_version",
            "--format=csv,noheader,nounits",
        ],
        timeout=60,
    )
    save("nvidia-smi.log", gpu)
    fields = [x.strip() for x in gpu.stdout.splitlines()[0].split(",")]
    if len(fields) < 5 or fields[1] != "Tesla T4" or fields[2] != "7.5":
        raise RuntimeError(f"requires Tesla T4 sm_75, got {fields}")

    phase = "reconstruct"
    clone = run(["git", "clone", "--filter=blob:none", REPO_URL, TREE], timeout=1800)
    save("git-clone.log", clone)
    checkout = run(["git", "checkout", "--detach", BASE_REV], cwd=TREE, timeout=600)
    save("git-checkout.log", checkout)
    applied = run(["git", "apply", "--whitespace=error", patch_path], cwd=TREE)
    save("git-apply.log", applied)
    diff = run(["git", "diff", "--check"], cwd=TREE)
    save("git-diff-check.log", diff)

    cargo_candidates = [
        shutil.which("cargo"),
        Path.home() / ".cargo/bin/cargo",
        "/usr/local/cargo/bin/cargo",
        "/opt/conda/bin/cargo",
    ]
    cargo = next((str(x) for x in cargo_candidates if x and Path(x).is_file()), None)
    cargo_env = {}
    bootstrapped = False
    if cargo is None:
        bootstrapped = True
        rustup_script = ROOT / "rustup-init.sh"
        urllib.request.urlretrieve("https://sh.rustup.rs", rustup_script)
        cargo_home = ROOT / "cargo-home"
        rustup_home = ROOT / "rustup-home"
        cargo_env = {"CARGO_HOME": cargo_home, "RUSTUP_HOME": rustup_home}
        install = run(
            [
                "bash",
                rustup_script,
                "-y",
                "--profile",
                "minimal",
                "--default-toolchain",
                "stable",
                "--no-modify-path",
            ],
            env=cargo_env,
            timeout=1800,
        )
        save("rustup-install.log", install)
        cargo = str(cargo_home / "bin/cargo")
    if not Path(cargo).is_file():
        raise RuntimeError(f"cargo unavailable after bootstrap: {cargo}")
    common = {**cargo_env, "CARGO_TARGET_DIR": TARGET, "CUDA_VISIBLE_DEVICES": "0"}
    (RESULTS / "cargo-discovery.json").write_text(
        json.dumps({"selected": cargo, "bootstrapped": bootstrapped}, indent=2),
        encoding="utf-8",
    )

    phase = "host-tests"
    tests = run([cargo, "test", "-p", "glcuda", "--lib", "--locked"], cwd=TREE, env=common)
    save("cargo-lib-tests.log", tests)
    if "67 passed" not in tests.stdout or "0 failed" not in tests.stdout:
        raise RuntimeError("unexpected host test summary")

    phase = "compiler-resource"
    ptxas = shutil.which("ptxas") or "/usr/local/cuda/bin/ptxas"
    resource = run(
        [
            ptxas,
            "-v",
            "-arch=sm_75",
            TREE / "glcuda/src/kernels/glcuda_sm75_wave88.ptx",
            "-o",
            ROOT / "wave125.cubin",
        ],
        check=False,
    )
    save("ptxas-wave125.log", resource)
    entry = resource.stderr
    registers = re.search(r"Used (\d+) registers", entry)
    if (
        resource.returncode
        or not registers
        or int(registers.group(1)) > 80
        or "0 bytes spill stores, 0 bytes spill loads" not in entry
        or "16384 bytes smem" not in entry
    ):
        raise RuntimeError(f"Wave 125 compiler resource gate failed: {entry}")
    (RESULTS / "wave125-resource.txt").write_text(entry, encoding="utf-8")

    phase = "cuda-parity"
    parity = run(
        [
            cargo,
            "test",
            "--release",
            "-p",
            "glcuda",
            "--test",
            "parity",
            "--locked",
            "--",
            "--nocapture",
            "--test-threads=1",
        ],
        cwd=TREE,
        env=common,
        check=False,
    )
    save("cargo-cuda-parity.log", parity)
    if parity.returncode or "0 failed" not in parity.stdout:
        raise RuntimeError("CUDA parity failed")

    phase = "build"
    build = run(
        [
            cargo,
            "build",
            "--release",
            "-p",
            "glcuda",
            "--example",
            "wave125_n16_fused_swiglu",
            "--example",
            "wave125_production",
            "--locked",
        ],
        cwd=TREE,
        env=common,
    )
    save("cargo-build.log", build)
    prod_env = {
        **common,
        "GLCUDA_FORCE_Q8": "1",
        "GLCUDA_GRID2D": "1",
        "GLCUDA_FUSE_Q8_GLUE": "1",
        "GLCUDA_Q8_NOSTORE": "1",
        "GLCUDA_FFN_GATE_UP_STACKED": "1",
        "GLCUDA_DEFER_FFN_RESIDUAL": "1",
        "GLCUDA_NTILE128": "1",
        "GLCUDA_BSTAGE": "1",
        "GLCUDA_GEMM_N16": "1",
        "GLCUDA_GEMM_N16_PREFETCH": "1",
        "GLCUDA_N16_FUSED_SWIGLU": "1",
        "GLCUDA_ATTN_MMA4": "1",
        "GLCUDA_ATTN_MMA4_REGQ": "1",
        "GLCUDA_ATTN_MMA4_AV": "1",
    }

    phase = "direct"
    direct_exe = TARGET / "release/examples/wave125_n16_fused_swiglu"
    direct = run([direct_exe], cwd=TREE, env=prod_env, check=False)
    save("wave125-direct.log", direct)
    direct_record = re.search(r"\[wave125-direct\]\s*(\{[^\n]+\})", direct.stdout)
    resource_record = re.search(r"\[wave125-resource\]\s*(\{[^\n]+\})", direct.stdout)
    if direct.returncode or not direct_record or not resource_record:
        raise RuntimeError("Wave 125 direct parity/resource gate failed")
    direct_json = json.loads(direct_record.group(1))
    occupancy_json = json.loads(resource_record.group(1))
    if not direct_json.get("q8_bit_exact") or not direct_json.get("scale_bit_exact"):
        raise RuntimeError(f"Wave 125 exact parity failed: {direct_json}")
    if occupancy_json.get("active_blocks_per_sm", 0) < 3:
        raise RuntimeError(f"Wave 125 occupancy failed: {occupancy_json}")

    phase = "model"
    urllib.request.urlretrieve(MODEL_URL, MODEL)
    model_meta = {"bytes": MODEL.stat().st_size, "sha256": sha256_file(MODEL)}
    if model_meta != {"bytes": MODEL_BYTES, "sha256": MODEL_SHA256}:
        raise RuntimeError(f"model identity mismatch: {model_meta}")
    (RESULTS / "model.json").write_text(json.dumps(model_meta, indent=2), encoding="utf-8")

    phase = "production-ab"
    prod_exe = TARGET / "release/examples/wave125_production"
    records = []
    invocation_summaries = []
    for invocation in ["a", "b"]:
        measured = run([prod_exe, MODEL, invocation], cwd=TREE, env=prod_env, check=False)
        save(f"production-{invocation}.log", measured)
        if measured.returncode:
            raise RuntimeError(f"production invocation {invocation} failed")
        records.extend(
            json.loads(raw)
            for raw in re.findall(r"\[wave125-sample\]\s*(\{[^\n]+\})", measured.stdout)
        )
        found = re.search(r"\[wave125-summary\]\s*(\{[^\n]+\})", measured.stdout)
        if not found:
            raise RuntimeError(f"production invocation {invocation} summary missing")
        invocation_summaries.append(json.loads(found.group(1)))
    if len(records) != 80 or {r["oracle_token"] for r in records} != {3323}:
        raise RuntimeError(f"production sample/oracle contract failed: {len(records)} records")
    retained = [r["prefill_ms"] for r in records if r["arm"] == "retained"]
    candidate = [r["prefill_ms"] for r in records if r["arm"] == "candidate"]
    if len(retained) != 40 or len(candidate) != 40:
        raise RuntimeError(f"wrong arm counts: retained={len(retained)} candidate={len(candidate)}")
    retained_median = statistics.median(retained)
    candidate_median = statistics.median(candidate)
    speedup = retained_median / candidate_median
    retained_tps = 244000.0 / retained_median
    candidate_tps = 244000.0 / candidate_median
    gate = {
        "minimum_speedup": 1.04,
        "combined_speedup_pass": speedup >= 1.04,
        "both_invocations_positive": all(x["speedup"] > 1.0 for x in invocation_summaries),
    }
    gate["passed"] = gate["combined_speedup_pass"] and gate["both_invocations_positive"]

    phase = "candidate-profile"
    profile = run(
        [prod_exe, MODEL, "profile"],
        cwd=TREE,
        env={**prod_env, "GLCUDA_TELEMETRY": "1"},
        check=False,
    )
    save("candidate-profile.log", profile)
    found_profile = re.search(r"\[wave125-profile\]\s*(\{[^\n]+\})", profile.stdout)
    if profile.returncode or not found_profile:
        raise RuntimeError("candidate profile failed")
    profile_json = json.loads(found_profile.group(1))
    stages = [
        json.loads(raw)
        for raw in re.findall(r"\[wave125-stage\]\s*(\{[^\n]+\})", profile.stdout)
    ]
    summary = {
        "wave": 125,
        "build": BUILD,
        "gpu": fields,
        "model": model_meta,
        "compiler": {"registers": int(registers.group(1)), "static_shared_bytes": 16384},
        "occupancy": occupancy_json,
        "direct": direct_json,
        "production_ab": {
            "samples_per_arm": 40,
            "retained_median_ms": retained_median,
            "candidate_median_ms": candidate_median,
            "retained_tps": retained_tps,
            "candidate_tps": candidate_tps,
            "speedup": speedup,
            "invocation_summaries": invocation_summaries,
        },
        "retention_gate": gate,
        "target_15000_tps_achieved": candidate_tps >= 15000.0,
        "profile": profile_json,
        "stages": stages,
        "glbench_missing_counters": [
            "eligible warps per cycle",
            "issue-active percentage",
            "stall-reason breakdown",
            "L1/L2 hit rate by kernel",
        ],
    }
    (RESULTS / "production-records.json").write_text(
        json.dumps(records, indent=2), encoding="utf-8"
    )
    (RESULTS / "wave125-summary.json").write_text(
        json.dumps(summary, indent=2), encoding="utf-8"
    )
    print("WAVE125_RESULT", json.dumps(summary, indent=2), flush=True)
    archive()
except Exception:
    (RESULTS / "FAILED.json").write_text(
        json.dumps({"phase": phase, "traceback": traceback.format_exc()}, indent=2),
        encoding="utf-8",
    )
    archive()
    raise
```
